```python
# Import library
from pyspark.sql import SparkSession
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    LongType,
    IntegerType
)
from pyspark.sql.functions import avg

# 1. Membuat SparkSession
spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

# 2. Mengambil seluruh data dari MongoDB
dokumen_menu = list(koleksi_menu.find({}))

print("Jumlah data dari MongoDB:", len(dokumen_menu))

# 3. Menyesuaikan struktur data
data_spark = []

for menu in dokumen_menu:
    data_spark.append({
        "_id": str(menu["_id"]),
        "nama_menu": menu.get("nama_menu"),
        "harga": int(menu["harga"]),
        "kategori": menu.get("kategori"),
        "tingkat_kepahitan": (
            int(menu["tingkat_kepahitan"])
            if menu.get("tingkat_kepahitan") is not None
            else None
        ),
        "pedas_level": (
            int(menu["pedas_level"])
            if menu.get("pedas_level") is not None
            else None
        ),
        "stok": (
            int(menu["stok"])
            if menu.get("stok") is not None
            else None
        ),
        "rekomendasi_barista": (
            str(menu["rekomendasi_barista"])
            if menu.get("rekomendasi_barista") is not None
            else None
        ),
        "status": menu.get("status")
    })

# 4. Menentukan skema Spark DataFrame
schema_menu = StructType([
    StructField("_id", StringType(), True),
    StructField("nama_menu", StringType(), True),
    StructField("harga", LongType(), True),
    StructField("kategori", StringType(), True),
    StructField("tingkat_kepahitan", IntegerType(), True),
    StructField("pedas_level", IntegerType(), True),
    StructField("stok", IntegerType(), True),
    StructField("rekomendasi_barista", StringType(), True),
    StructField("status", StringType(), True)
])

# 5. Mengubah data menjadi Spark DataFrame
df_spark = spark.createDataFrame(
    data_spark,
    schema=schema_menu
)

# 6. Menampilkan struktur dan isi DataFrame
print("\nStruktur Spark DataFrame:")
df_spark.printSchema()

print("\nData katalog Kopi Nusantara:")
df_spark.show(12, truncate=False)

# 7. Menghitung rata-rata harga berdasarkan kategori
print("\nRata-rata harga berdasarkan kategori:")

hasil_analisis = df_spark.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga")
)

hasil_analisis.show()

# 8. Menutup SparkSession
spark.stop()

print("SparkSession berhasil ditutup.")
```

Melalui pengerjaan tugas mandiri ini, saya menemukan bahwa salah satu tantangan dalam mengintegrasikan MongoDB dengan PySpark adalah perbedaan struktur data yang dimiliki setiap dokumen. MongoDB memungkinkan setiap dokumen memiliki atribut yang berbeda, sedangkan Spark DataFrame membutuhkan struktur kolom dan tipe data yang seragam. Sebagai contoh, menu kopi memiliki field tingkat_kepahitan, sementara menu makanan memiliki field pedas_level. Untuk mengatasi hal tersebut, saya menggunakan metode get() agar setiap dokumen tetap memiliki susunan field yang sama. Jika suatu field tidak tersedia, saya memberikan nilai None sehingga dapat direpresentasikan sebagai null pada Spark DataFrame. Selain itu, saya juga menentukan skema secara eksplisit menggunakan StructType dan StructField agar tipe data setiap kolom dapat dikenali dengan baik oleh Spark. Melalui proses ini, saya memahami bahwa integrasi antara MongoDB dan PySpark membutuhkan penyesuaian struktur data sebelum proses analisis dilakukan.
